```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
## let's create synthetic data points
from sklearn.datasets import make_classification
```

```python
x,y = make_classification(n_samples=1000,n_features=2,n_classes=2,n_clusters_per_class=1,n_redundant=0)
```

```python
x,y
```

```python
df  =pd.DataFrame(x)
```

```python
sns.scatterplot(x=df[0],y=df[1],hue=y)
# this is a graph of 2 featues dataset and the out is either 0 or 1 
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(x,y,test_size=0.25,random_state=10)
```

```python
from sklearn.svm import SVC
```

```python
svc = SVC(kernel='linear')
```

```python
svc.fit(X_train,Y_train)
```

```python
# prediction 
Y_pred = svc.predict(X_test)
```

```python
Y_pred
```

```python
from sklearn.metrics import classification_report, confusion_matrix
```

```python
print(classification_report(Y_test,Y_pred))
print(confusion_matrix(Y_test,Y_pred))
# high accuracy because very limited points are overlapping
```

```python
# for very much overlapping points we can use 'rbf' kernal. 
# can use multiple kernals 
```

```python
sigmoid = SVC(kernel='sigmoid')
sigmoid.fit(X_train,Y_train)
# prediction 
Y_pred1 = sigmoid.predict(X_test)
print(classification_report(Y_test,Y_pred1))
print(confusion_matrix(Y_test,Y_pred1))
```

```python
# here it get lower accuracy according to the older kernal used.
```

### Hyperparametric training

```python
# now we try to apply gridsarchCV here with kernals
from sklearn.model_selection import GridSearchCV

param_grid = {
    'C':[0.1,1,10,100,1000],
    'gamma':[1,0.1,0.01,0.001,0.0001],
    'kernel':['rbf']
}
```

```python
grid = GridSearchCV(SVC(),param_grid=param_grid,cv=5,verbose=3,refit=True)
```

```python
grid.fit(X_train,Y_train)
```

```python
grid.best_params_
```

```python
Y_pred3 = grid.predict(X_test)
print(classification_report(Y_test,Y_pred3))
print(confusion_matrix(Y_test,Y_pred3))
```
